## Function

### Functions in Python are reusable blocks of code that perform a specific task and run only when called. They help break complex programs into smaller, organized pieces so you do not have to repeat the same code.

```python
#Create a function for num that is odd or even
def num(i):
    """ This fuction is display the num is odd or Even"""
    if i%2 ==0 :
        return "is Even"
    else:
        return "not Even"
```

```python
for i in range(1,11):
    x = num(i)
    print(x)
```

### Types of arguments

```python
# Default argument
# Positional argument
# Keyword argument
```

```python
# Default argument
def multi(a=7,b=12):
    return a*b
print(multi())
```

```python
# Positional argument
print(multi(5,9))
```

```python
# Keyword argument
print(multi(b=5,a=7))
```
