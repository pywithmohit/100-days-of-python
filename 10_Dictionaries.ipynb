## Dictonary 

### Dictonary is used to store data in key:value pair. Dictonary items are ordered ,changeable, and do not allows duplicated values.

```python
dict1 = {"name":"Mohit",
         "age":25,
         "subject":"math"
         }
print(type(dict1))
```

### accessing items

```python
# []
print(dict1["name"])
#get
print(dict1.get("age")) 
```

### adding key values pair

```python
dict1["class"] = "Fifth"
print(dict1) #add key pair
```

### Remove key items

```python
dict1.pop("age")
print(dict1) #delete age key pair
```

```python
#popitem
dict1.popitem() #its delete last item or dictionary
```

```python
# del
del dict1['subject']
print(dict1) 
```
