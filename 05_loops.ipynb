## loops in python

### loops are set of instuction that can be use to execute block of code repeatedly.

### Loops types -->

```python
# while loop example to print a table.
num = int(input("Enter the number --> "))
i = 1
while i < 11:
   print(i,"*", num ,"=" ,i*num)
   i += 1
```

```python
# Iterate 0 to 10 using for loop, do the same using while loop.

i = 0
while i < 11:
    print(i)
    i += 1
```

```python
# Iterate 10 to 0 using for loop, do the same using while loop.
i = 10
while i > -1:
    print(i)
    i -= 1
```

```python
# Write a loop that makes seven calls to print(), so we get on the output the following triangle:
  #
  ##
  ###
  ####
  #####
  ######
  #######

num = "#"
i = 1
while i < 8:
    print(num*i)
    i += 1
```

```python
# Use nested loops to create the following:
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
row = 8
col = 8
for i in range(row):
    for j in range(col):
     print("#",end="")
    print()
```

```python
# Iterate through the list, ['Python', 'Numpy','Pandas','Django', 'Flask'] using a for loop and print out the items.
lang = ['Python', 'Numpy','Pandas','Django', 'Flask']
for i in lang:
    print(i)
```

```python
# Use for loop to iterate from 0 to 100 and print only even numbers
for i in range(1,101):
    if i % 2 == 0:
        print(i)
```

```python
# Use for loop to iterate from 0 to 100 and print only odd numbers
for i in range(1,101):
    if i % 2 != 0:
        print(i)
```

```python
# Use for loop to iterate from 0 to 100 and print the sum of all numbers.
sum = 0
for i in range(1,101):
     sum += i
print(sum)
```

```python
# Use for loop to iterate from 0 to 100 and print the sum of all evens and the sum of all odds.
sum = 0
for i in range(1,101):
    if i % 2 == 0:
     sum += i
print(sum)
```

```python
# Random Guessing Game
import random
jackpot = random.randint(1,5)
count = 0


while True:
    count += 1
    guess = int(input("Enter the number"))
    if guess < jackpot:
        print("Guess a higher number")
    elif guess > jackpot:
        print("Guess a lower number")
    else:
        print("Yehh Correct guess")
        break
```
