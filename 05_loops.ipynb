## loops in python

### loops are set of instuction that can be use to execute block of code repeatedly.

### Loops types -->

#### 1. for loop
#### 2. while loop

```python
# while loop example to print a table.
num = int(input("Enter the number --> "))
i = 1
while i < 11:
   print(i,"*", num ,"=" ,i*num)
   i += 1
```

```python
# Iterate 0 to 10 using for loop, do the same using while loop.

i = 0
while i < 11:
    print(i)
    i += 1
```

```python
# Iterate 10 to 0 using for loop, do the same using while loop.
i = 10
while i > -1:
    print(i)
    i -= 1
```

```python
# Write a loop that makes seven calls to print(), so we get on the output the following triangle:
  #
  ##
  ###
  ####
  #####
  ######
  #######

num = "#"
i = 1
while i < 8:
    print(num*i)
    i += 1
```
