## Strings

### Strings are sequence of characters

```python
# creating strings 
name = "Mohit yadav"
city = 'Jaipur'
address = """"139/368 Sahu nagar sawai madhopur 
                 rajstan east india """ #Triple inverted commas are useed for multiline strings
print(name)
print(city)
print(address)
```

```python
# Access substrings from strings
print(name[::]) #slicing
print(city[2:8]) #slicing
print(address[:11:4]) #slicing
print(city[-2]) #indexing
```

## Count/find/index

```python
text = "banana bandana"

# count returns the number of non-overlapping matches
print(text.count("ana"))

# find returns the position of the first match, or -1 if it is missing
print(text.find("band"))
print(text.find("xyz"))

# index also returns the position, but raises ValueError if the text is missing
print(text.index("band"))
```

## Endswith/Startwith

```python
#check the string ends with the statment
data = "Found a negociated"
print(data.endswith("m")) # false output
print(data.endswith("ted")) #True output
```

```python
#check the string start with the statment
print(data.startswith("neg"))
print(data.startswith("Fou"))
```
