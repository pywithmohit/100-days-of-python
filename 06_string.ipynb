## Strings

### Strings are sequence of characters

```python
# creating strings 
name = "Mohit yadav"
city = 'Jaipur'
address = """"139/368 Sahu nagar sawai madhopur 
                 rajstan east india """ #Triple inverted commas are useed for multiline strings
print(name)
print(city)
print(address)
```

```python
# Access substrings from strings
print(name[::]) #slicing
print(city[2:8]) #slicing
print(address[:11:4]) #slicing
print(city[-2]) #indexing
```

## editing or deleting a string

```python
# strings in python is immuteble
```

```python
# arithimetic operation in strings 
print(name + ' ' + city)
print(city*4)

```

```python
#logical operator 
print("delhi" == "delhi")
print("delhi" != "delhi")
print('mumbai'>'june') #lexicographical (alphabetical) comparison.
```

```python
'hello' and 'world' #Returns the second value because the first is true.
```

```python
'' and "World" #Returns the 1st value because the first is false.
```

```python
'hello' or 'world' #Short-circuits on or because the first value is already true.
```
