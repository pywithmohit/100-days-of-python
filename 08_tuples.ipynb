## Tuples

### A tuples is in python similar to the list. the difference is we cannot change the elements of a tuples once it assigned.
### in short tuples are immutable list.

## characterstics

- Ordered 
-  Unchangeable
- Allows Duplicate

```python
# creating tuples
a = ("Ram",25,"Akash",True)
print(type(a),a)
```

```python
# creating tuples with single item
t1 = (2,) # use comma for single tuple item
print(type(t1))
```

###  Accessing item

```python
# 1.Slicing
print(a[2:4])
print(a[2:-1])
print(a[:])
print(a[::-1]) #reverse the set item
```

```python
# 2. Indexing
print(a[0]) #positive indexing 
print(a[-1]) # negative indexing
```

### Edit tuples item

```python
data = ("Ram","raju","Akash","Priya")
data[2] = "Ashi" #display error because tuples are immutable
```
