## List in python

### list is a data type where you can store multiple types of data.technically list act like dynamic array which mean you can add data in a single container.


```python
a = ["jessica",25,88.55]
print(a)
print(id(a[2]))
```

 ## Charachterstics of List

```python
# Demonstrating characteristics of a list using the existing variable 'a'

print("Original list:", a)
print("Type of a:", type(a))
print("Length of a:", len(a))

# 1. Lists are ordered
print("Ordered access:")
print("First element:", a[0])
print("Second element:", a[1])
print("Third element:", a[2])

# 2. Lists are mutable
print("Before modification:", a)
a[1] = 30
print("After modification:", a)

# 3. Lists can store multiple data types
print("Mixed data types:", a)

# 4. Lists can be extended
a.append("Python")
print("After append:", a)

# 5. Lists allow duplicates
dup_list = [1, 2, 2, 3, 3]
print("Duplicate values are allowed:", dup_list)
```

```python
# Creating a new list
new_list = ["apple", "banana", "cherry", 42, 3.14]

print("Created list:", new_list)
print("Type:", type(new_list))
print("Length:", len(new_list))
print("First item:", new_list[0])
print("Last item:", new_list[-1])
```

```python
# Access list items using index
print("a[0]:", a[0])
print("a[1]:", a[1])
print("a[-1]:", a[-1])  # last item

# Access a range of items
print("a[0:3]:", a[0:3])
print("new_list[:3]:", new_list[:3])
print("new_list[-2:]:", new_list[-2:])

# Iterate through a list
print("Items in dup_list:")
for item in dup_list:
    print(item)
```

```python
# append() adds one item to the end
new_list.append("orange")
print("After append:", new_list)

# insert() adds one item at a specified index
new_list.insert(1, "mango")
print("After insert:", new_list)

# extend() adds multiple items to the end
new_list.extend(["grape", "kiwi"])
print("After extend:", new_list)
```

```python
# Edit a specific item in an existing list
a[0] = "Jessica"  # replace the first element
print("Updated a:", a)


# Edit multiple list items using slicing
new_list[0:2] = ["apple", "blueberry"]
print("Updated new_list:", new_list)
```

## Deleting item in List

```python
fruits = ["apple","Banana","Grapes","Orange","Lemon"]


```

```python
fruits.remove("apple") #remove by particular name
print(fruits)
```

```python
fruits.pop(2) #remove item using index
print(fruits)
```

```python
fruits.clear() # empty all the list items
print(fruits)
```
