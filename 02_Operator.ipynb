# Operator in python

### 1. arithmetic operator
### 2. relational operator
### 3. logical operator
### 4. Bitwise operator
### 5. assignment opretor
### 6. membership operator

```python
# Airthmetic Operator

print(500+55)
print(55*88)
print(88/5)
print(77//5)
print(5%2)
```

```python
# Relational Operator

print(8 < 5)
print(8==5)
print(55 > 55)
print(9>=7)
```

```python
# Logical Operator

print(4 or 3)
print(1 and 5)
print(not 1)
```

```python
# Bitwise Operator
print(6 & 2)  #AND
print(6 | 8)  #OR
print(6 ^ 3)  #XOR
```

```python
# Assignment Operator --> Assignment operators are used to assign values to variables
a = 45
a += 45
print(a)
```

```python
# Membership Operator --> Membership operators are used to test if a sequence is presented in an object
fruits = ["Apple","Banana","Grapes","Gauava"]
print("Apple" in fruits)
print("Banana" not in fruits)
```
